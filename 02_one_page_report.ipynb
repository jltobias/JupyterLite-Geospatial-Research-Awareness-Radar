# Geospatial Research Radar - 02: one-page report

This notebook turns the latest morning scan into two compact artifacts:

1. A **clickable one-page PDF** generated with no third-party Python package.
2. A **Unicode HTML version** that is easy to print to PDF from the browser and is more tolerant of non-Latin article titles.

The report shows only the top-ranked items so the output stays scan-friendly rather than becoming another feed.

```python
from pathlib import Path
import json
from radar_core import load_json, create_one_page_pdf, create_one_page_html, DEFAULT_CONFIG

files = sorted(Path('.').glob('radar_results_*.json'))
if not files:
    raise FileNotFoundError("No radar_results_YYYY-MM-DD.json found. Run 01_morning_scan.ipynb first.")

latest = files[-1]
data = load_json(latest)
items = data.get('items', [])
run_date = data.get('run_date', latest.stem.replace('radar_results_', ''))
max_items = load_json('radar_config.json', DEFAULT_CONFIG).get('max_report_items', 7)

pdf_path = f"geospatial_radar_{run_date}.pdf"
html_path = f"geospatial_radar_{run_date}.html"

create_one_page_pdf(items, pdf_path, run_date=run_date, max_items=max_items)
create_one_page_html(items, html_path, run_date=run_date, max_items=max_items)

print("Created:", pdf_path)
print("Created:", html_path)
print("Items in brief:", min(len(items), max_items))
```

## Optional inline preview

JupyterLite can open the HTML artifact directly from its file browser. In a normal Jupyter environment, the cell below also displays the report inline.

```python
try:
    from IPython.display import HTML, display
    display(HTML(Path(html_path).read_text(encoding='utf-8')))
except Exception as e:
    print("Open", html_path, "from the JupyterLite file browser.")
```
