The starter profile emphasizes public-health GIS, spatial epidemiology, geoprivacy, GeoAI/geospatial foundation models, population mapping/mobility, Earth observation, spatial statistics, and open/reproducible Python workflows.

```python
from radar_core import DEFAULT_CONFIG, deep_copy_jsonable, save_json, make_external_search_queries

config = deep_copy_jsonable(DEFAULT_CONFIG)
config["lookback_days"] = 4
config["max_results_per_query"] = 12
config["max_report_items"] = 7
config["minimum_priority_score"] = 42

# Optional API keys. OpenAlex supports casual keyless use; a free key gives more budget.
# Semantic Scholar also supports an optional key.
config["api_keys"]["openalex"] = ""
config["api_keys"]["semantic_scholar"] = ""

save_json(config, "radar_config.json")
print("Saved radar_config.json")
```

## Tune the score

The default score has three layers:

- **Interest score (0-100):** independent topic groups + watched-author bonus + recency + open-access signal.
- **MVP score (0-100):** open-data signals + implementation/code signals + concrete methods + small-compute fit.
- **Priority score:** `0.58 * interest + 0.42 * MVP`.

Edit the term lists in `radar_config.json` or in `radar_core.py` if the radar becomes too broad or too narrow.
